```python
import numpy
from numba import types, njit
from typing import Tuple


@njit    
def location_add_numba(loc: Tuple[int, int], direction: Tuple[int, int]) -> Tuple[int, int]:
    return (loc[0] + direction[0], loc[1] + direction[1])

def location_add(loc: Tuple[int, int], direction: Tuple[int, int]) -> Tuple[int, int]:
    return (loc[0] + direction[0], loc[1] + direction[1])



def no_numba():
    for i in range(100):
        location_add((0,1), (0, 1))


@njit  
def with_numba():
    for i in range(100):
        location_add_numba((0, 1),(0, 1))

num_runs = 200000  # number of repetitions (100k is good for small data)
import timeit
#Numba speed
no_numba_time = timeit.timeit(
    no_numba,
    globals=globals(), 
    number=num_runs)

#
with_numba_time = timeit.timeit(
    with_numba, 
    globals=globals(), 
    number=num_runs)


# --- Results ---
print(f"100x Direction Add runs: {num_runs:,}\n")
print(f"No Numba           : {no_numba_time:.6f} s total → {no_numba_time/num_runs:.3e} s per copy")
print(f"With Numba            : {with_numba_time:.6f} s total → {with_numba_time/num_runs:.3e} s per copy")
```
