```python
import numpy as np
from numba import njit

arr = np.zeros((32, 32), dtype=np.uint8)

l = list()
s = set()

for i in range(900):
    rand_x = np.random.randint(0, 32)
    rand_y = np.random.randint(0, 32)
    l.append((rand_x,rand_y))
    s.add((rand_x,rand_y))

    arr[rand_y][rand_x] = 1

@njit
def numba_copy(a):
    return a[:, :]

# Warm-up JIT compilation
_ = numba_copy(arr)
```
